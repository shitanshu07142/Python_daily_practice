Exception Handling

```python
def get_valid_id():
    while True:
        try:
            employee_id=int(input("please enter your employee id: "))
            return employee_id
        except ValueError:
            print("invalid input")
emp_id=get_valid_id()
print("Your employee id is: ",emp_id)
        
```

```python
def get_valid_salary():
    while True:
        try:
            salary=float(input("Enter Monthly Salary: "))
            if salary<0:
                print("Salary must be greater than zero!")
            else:
                return salary
        except ValueError:
            print("invalid input")
            
emp_salary=get_valid_salary()
print("Employee Salary is : ",emp_salary)
```

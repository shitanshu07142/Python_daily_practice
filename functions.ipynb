```python
def calculate_bonus():
    salary=int(input("please enter your salary: "))
    rating=int(input("please tell me rating: "))
    if rating>=4:
        bonus=salary*0.15
    elif rating==3:
        bonus=salary*0.10
    else:
        bonus=0
    return bonus,salary
calculate_bonus()
```

```python
employees = [
    (101, "Aman", "HR", 45000.0),
    (102, "Priya", "IT", 65000.0),
    (103, "Kunal", "Finance", 52000.0)
]
def display_all_records(records_list):
    for emp_id,name,dept,salary in records_list:
        print(f"{emp_id} {name}|Depatment:{dept}|Salary:Rs{salary} ")
        print("-----------------------------------------------")
display_all_records(employees)
```

```python
employees = [
    (101, "Aman", "HR", 45000.0),
    (102, "Priya", "IT", 65000.0),
    (103, "Kunal", "Finance", 52000.0)
]
def find_employee(emp_list,target_id):
    for employee in employees:
        if employee[0]==target_id:
            return employee
    return None
find_employee(employees,103)
```

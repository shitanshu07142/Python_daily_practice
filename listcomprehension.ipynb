List Comprehension

```python
data=[1,2,3,4,5]
result=[]
for x in data:
    result.append(x**2)
print(data)
print(result)
```

```python
data=[1,2,3,4,5]
result=[x**2 for x in data]
print(result)
```

```python
data=[1,2,3,4,5]
result=[x**2 for x in data if x%2==0]
print(result)
```

```python
marks=[60,44,78,98,4,34,65]
result=["pass" if ex>=60 else "fail" for ex in marks]
print(result)
```

```python
numbers=[10,25,35,60,45,15,50,70]
even=[x for x in numbers if x%2==0]
print(even)
```

```python
Names=["amit","soniya","aditi","abhay","adarsh"]
upper=[x.upper() for x in Names]
print(upper)
```

```python
Names=[" amit "," soniya "," aditi "," abhay ","adarsh"]
result=[x.strip() for x in Names]
print(result)
```

```python
numbers=[20,55,30,80,45,90,10]
great=[x for x in numbers if x>50]
print(great)
```

```python
numbers=[10,15,20,30,45,15,50,70]
between=[x for x in numbers if x>=20 and x<=50]
print(between)
```

```python
numbers=[10,15,20,30,45,50,60,75]
devide=[x for x in numbers if x%3==0 and x%5==0]
print(devide)
```

```python
names=["amit","rahul","neha","rohit","sneha"]
char=[x for x in names if len(x)>=4]
print(char)
```

```python
names=["amit","rahul","neha","rohit","sneha"]
char=[x for x in names if x.startswith("a")]
print(char)
```

```python
text="programming"
vowel=[x for x in text if x in "aeiou"]
print(vowel)
```

```python
numbers=[10,15,20,25,30]
oddoreven=["even" if x%2==0 else "odd" for x in numbers ]
print(oddoreven)
```

```python
numbers=[10,-5,20,-8,30,-2]
positive=[0 if x<0 else x for x in numbers]
print(positive)
```

```python
prices=[500,1200,800,2000,1500]
newprice=[x-(x/10) if x >1000 else x for x in prices]
print(newprice)
```

```python
sales=[["aman",12000,15000,18000],
      ["priya",20000,17000,14000],
      ["rohit",16009,22000,19000]]
highestsale=0
top_salesman=""
for salesman in sales:
    name=salesman[0]
    total=sum(salesman[1:])
    print("Name : ",name)
    print("total sales : ",total)
    print("-------------------------")
    if total>highestsale:
        highestsale=total
        top_salesman=name
print("highest sale: ",highestsale)
print("name: ",top_salesman)
```

```python
def safe_int(val,default=0):
    try:
        int(val)
        return int(val)
    except (TypeError,ValueError):
        return default
print(safe_int("42"))
print(safe_int("hello",-1))
print(safe_int(None,0))
print(safe_int([1,2,3]))
```

```python
def calculate(operation,*numbers):
    if len(numbers)==0:
        raise ValueError("No numbers provides")
    elif operation=="add":
        add=sum(numbers)
        return add
    elif operation=="multiply":
        total=1
        for i in numbers:
            multiply=total*i
        return multiply
    elif operation=="average":
        average=sum(numbers)/len(numbers)
        return average
    else :
        raise ValueError("Unknown operation: subtract")
print(calculate("add",1,2,3,4))
print(calculate("multiply",1,2,3,4))
print(calculate("average",15,24,37,42))
#calculate("subtract",5,2)
#calculate("add")
```

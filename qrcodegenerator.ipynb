```python

from IPython.display import display
url=input("enter url to make QR : ")

if url.startswith("http://") or url.startswith("https://"):
    qr=qrcode.make(url)
    display(qr)
else:
    print("invalid url must start with http or https")
```

```python
from IPython.display import display
url=input("enter url to make QR : ")
if url.startswith("https://") or url.startswith("https://"):
    qr=qrcode.QRCode(version=1,
    error_correction=qrcode.constants.ERROR_CORRECT_H,box_size=10,
                     border=4)
    qr.add_data(url)
    qr.make()
    image=qr.make_image(fill_color="white",back_color="black")
else:
    print("invalid url must start with http or https")
display(image)
```

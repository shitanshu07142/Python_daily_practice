```python
from IPython.display import display
from PIL import Image,ImageDraw,ImageFont
txt=input("Please enter your text: ")
image=Image.new("RGB",(800,300),"white")
font=ImageFont.load_default()
draw = ImageDraw.Draw(image)
draw.text( (50,50) ,txt,fill="black",font=font)
display(image)
```

```python
from IPython.display import display
from PIL import Image,ImageDraw,ImageFont
text=input("Please enter your text: ")
image=Image.new("RGB",(800,400),"white")
draw = ImageDraw.Draw(image)
font=ImageFont.load_default()
left_margin=50
right_margin=50
available_width=800-left_margin-right_margin
words=text.split()
wrapped_lines=[]
current_line=""
for word in words:
    test_line=current_line + " " + word if current_line else word
    text_width=draw.textbbox((0,0), test_line, font=font)[2]
    if text_width<=available_width:
        current_line=test_line
    else :
        if current_line:
            wrapped_lines.append(current_line)
        current_line=word
if current_line:
    wrapped_lines.append(current_line)
y=50
for line in wrapped_lines:
    draw.text((left_margin,y),line,fill="red",font=font)
    y+=20
display(image)
```
